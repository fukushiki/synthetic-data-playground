### Get Formatted Phone Number
If you want to get a formatted phone number, you can use the `phone_number` method,
and create a formatted phone number like the following examples:

```python
import re

phone = fake.basic_phone_number()
digits = re.sub(r"\D", "", phone)

formatted_phone = f"{digits[:3]}-{digits[3:6]}-{digits[6:]}"

print(formatted_phone)
```

### seed

```python
from faker import Faker

fake_a = Faker("en_US")
fake_b = Faker("en_US")

# Set an independent random seed for each instance.
fake_a.seed_instance(42)
fake_b.seed_instance(42)

# Generate names independently.
name_a = fake_a.name()
name_b = fake_b.name()

assert name_a == name_b

print(name_a)
print(name_b)
```

```python
from faker import Faker

SEED = 42

fake_a = Faker("ja_JP")
fake_b = Faker("ja_JP")

# seed_instance()でSEED値を固定
fake_a.seed_instance(SEED)
fake_b.seed_instance(SEED)

# それぞれのインスタンスで連続して名前を5件生成
names_a = [fake_a.name() for _ in range(5)]
names_b = [fake_b.name() for _ in range(5)]

# 生成結果が一致することを確認
assert names_a == names_b

for i, (a, b) in enumerate(zip(names_a, names_b), start=1):
    print(f"{i}: fake_a={a}, fake_b={b}")
```

```python
fake_a = Faker("ja_JP")
fake_a.seed_instance(SEED)
print(fake_a.name())
print(fake_a.name())

print("seed_instanceを再定義")

fake_a.seed_instance(SEED)
print(fake_a.name())
print(fake_a.name())
print(fake_a.name())
```
