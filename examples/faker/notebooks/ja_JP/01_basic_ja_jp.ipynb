# 日本語Local生成

## インスタンス宣言とLocale

```python
from faker import Faker

fake_ja = Faker('ja_JP')
```

```python
fake_ja.name()
```

## Providerで提供されるメソッド

### Person

```python
# 日本語フルネーム
print(fake_ja.name()) # 例) '佐藤 さゆり'
```

```python
# 日本語 姓/名
print(fake_ja.first_name()) # 例) `さゆり`
print(fake_ja.last_name()) # 例) `佐藤`
```

```python
# First Name(名)は性別によくある名前を生成可能
print(fake_ja.first_name_male()) # 例) `学`
print(fake_ja.first_name_female()) # 例) `さゆり`
```

```python
# カタカナ名フルネーム 性別によくある名前を生成可能
print(fake_ja.kana_name()) # 例) `サトウ サユリ`
print(fake_ja.kana_name_male()) # 例) `サトウ サユリ`
print(fake_ja.kana_name_female()) # 例) `サトウ マナブ`

# カタカナ姓名
print(fake_ja.first_kana_name()) # 例) `サユリ`
print(fake_ja.last_kana_name()) # 例) `サトウ`

# カタカナFirstName(名)は性別によくある名前を生成可能
print(fake_ja.first_kana_name_male()) # 例) `マナブ`
print(fake_ja.first_kana_name_female()) # 例) `サユリ`
```

```python
# ローマ字でのフルネーム
print(fake_ja.romanized_name()) # 例) 'Sayuri Sato'
# ローマ字でのフルネーム 性別によくある名前を生成可能
print(fake_ja.romanized_name_male()) # 例) `Manabu Sato`
print(fake_ja.romanized_name_female()) # 例) 'Sayuri Sato'
```

```python
# Tupple形式の(漢字, カナ, ローマ字)生成
print(fake_ja.first_name_pair()) # 例) ('小百合', 'サユリ', 'Sayuri')
print(fake_ja.last_name_pair()) # 例) ('佐藤', 'サトウ', 'Sato')
# Tupple形式の(漢字, カナ, ローマ字)生成 性別によくある名前を生成可能
print(fake_ja.first_name_male_pair()) # 例) ('小百合', 'サユリ', 'Sayuri')
print(fake_ja.first_name_female_pair()) # 例) ('学', 'マナブ', 'Manabu')
```
