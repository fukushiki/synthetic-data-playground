# データサンプル生成 Case2-販売データ

## インスタンス宣言とLocale

```python
import pandas as pd
from faker import Faker

fake_ja = Faker("ja_JP")
fake_ja.seed_instance(42)
```

```python
import random
import pandas as pd
from datetime import date
from faker import Faker

fake_orders = Faker("ja_JP")
fake_orders.seed_instance(27)

rng = random.Random(27)
```

```python
customers = []

for i in range(1, 11):
    customers.append({
        "customer_id": f"C{i:04d}",
        "name": fake_ja.unique.name(),
        "email": fake_ja.unique.safe_email(),
        "phone": fake_ja.unique.phone_number(),
        "address": fake_ja.unique.address().replace("\n", " "),
        "registered_at": fake_ja.date_between(
            start_date="-1y",
            end_date="today"
        )
    })

customers_df = pd.DataFrame(customers)

display(customers_df)
```

```python
products = [
    {"name": "ノートPC", "price": 120000},
    {"name": "マウス", "price": 3000},
    {"name": "キーボード", "price": 8000},
    {"name": "モニター", "price": 35000},
    {"name": "USBハブ", "price": 2500},
]

orders = []
customers = customers_df.to_dict("records")

for i in range(1, 51):
    customer = rng.choice(customers)
    product = rng.choice(products)
    quantity = rng.randint(1, 3)

    orders.append({
        "order_id": f"O{i:05d}",
        "customer_id": customer["customer_id"],
        "order_date": fake_orders.date_between_dates(
            date_start=customer["registered_at"],
            date_end=date.today()
        ),
        "product_name": product["name"],
        "quantity": quantity,
        "unit_price": product["price"],
        "total_amount": product["price"] * quantity
    })

orders_df = pd.DataFrame(orders)

display(orders_df)
```
